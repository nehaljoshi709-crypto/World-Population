# EDA 

## Data Visualization
This section contains a series of visualizations to explore population trends, growth, density, and projections.

```python
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
print(df["Population (2024)"].dtype)
```

### Top 10 countires population according to year 2024

```python
top10=df.sort_values(by="Population (2024)",ascending=False).head(10)
```

```python
fig, ax = plt.subplots(figsize=(10,7))

sns.set_theme()
sns.barplot(top10, x=top10["Country"],y=top10["Population (2024)"], hue=top10["Country"],ax=ax)
ax.bar_label(ax.containers[0], fontsize=10)
plt.title("Top 10 Countries by Population (2024)")
plt.xlabel("Country")
plt.ylabel("Population")

plt.xticks(rotation=45)
#plt.tight_layout()

plt.show()
```

**Insights**

* India has the highest population in 2024.
* China is the second most populated country.
* The population drops significantly after the top two countries.
* Countries like the USA, Indonesia, Pakistan, and Nigeria also have very large populations.
* The top 10 countries contain a large share of the world's total population.

## Lowest Population

```python
top10=df.sort_values(by="Population (2024)",ascending=True).head(10)
```

```python
lowest10 = df.sort_values("Population (2024)").head(10)

plt.figure(figsize=(10,6))
plt.barh(lowest10["Country"], lowest10["Population (2024)"], color="skyblue")
plt.title("Lowest 10 Population Countries")
plt.xlabel("Population")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

**Insights**

* These countries have very small populations compared to the rest of the world.
* Most are island nations or very small countries.
* Population size is not directly related to land area.
* These countries contribute only a tiny fraction of the world's population.

## Highest Density:

```python
print(df["density_per_km_square"].dtype)
```

```python
df["density_per_km_square"]=pd.to_numeric(df["density_per_km_square"],errors="coerce")
```

```python
density=df.sort_values("density_per_km_square",ascending=False).head()
```

```python
density = df.sort_values(by="density_per_km_square", ascending=False).head(10)

print(density[["Country", "density_per_km_square"]])

import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.bar(density["Country"], density["density_per_km_square"])
plt.xticks(rotation=45)
plt.title("Top 10 Highest Density Countries")
plt.xlabel("Country")
plt.ylabel("Density (P/Km²)")
plt.tight_layout()
plt.show()
```

**Insights**

* Some countries have extremely high population density despite having small land areas.
* High density means many people live in a limited space.
* Dense countries often face challenges like housing shortages, traffic congestion, and pressure on resources.
* Population density varies much more than total population.

```python
top10_density = df.sort_values(by="density_per_km_square", ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.plot(top10_density["Country"],
         top10_density["density_per_km_square"],
         marker='o')

plt.title("Top 10 Highest Population Density Countries")
plt.xlabel("Country")
plt.ylabel("Population Density (P/Km²)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(True)
plt.show()
```

**Insights**

* The graph shows a sharp decline from the highest-density country to the lower-ranked countries.*
* Only a few countries have exceptionally high population density.
* Population density is unevenly distributed across countries.
* The trend highlights large differences in how crowded countries are.

```python
top10=df.sort_values("Population (2024)").head(10)
```

```python
plt.figure(figsize=(10,6))
plt.plot(top10["Country"], top10["Population (2024)"], marker='o')
plt.title("Top 10 Countries by Population")
plt.xlabel("Country")
plt.ylabel("Population")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

# Top 5 Highest Population Density Countries 

```python
top5_density = df.sort_values(by="density_per_km_square", ascending=False).head(5)

fig = px.pie(
    top5_density,
    values="density_per_km_square",
    names="Country",
    title="Top 5 Highest Population Density Countries"
)

fig.show()
```

## Insights:

* Macau has the highest population density, contributing about 23.7% of the total among the top five countries.
    
* Monaco is the second most densely populated country, with approximately 22.6%.
    
* Singapore ranks third, contributing around 19.5%.
* Hong Kong is fourth, with about 17.1%.
* Gibraltar is fifth, contributing approximately 17.0%.
* The percentage difference between these countries is relatively small, indicating that all five have very high population densities.
    
* These countries are mostly small in land area, which is one of the main reasons for their extremely high population density.
* High population density often results in limited land availability, high-rise buildings, efficient public transport, and well-developed urban infrastructure.

```python
import plotly.express as px

fig = px.histogram(
    df,
    x='Population (2024)',
    nbins=30,
    title='Population Distribution of Countries',
    labels={'Population (2024)': 'Population (2024)'},
)

fig.update_layout(
    xaxis_title='Population (2024)',
    yaxis_title='Number of Countries',
    template='plotly_white'
)

fig.show()
```

## Insights:
* Most countries have relatively small populations, as the tallest bars are concentrated on the left side of the graph.
* Only a few countries have extremely large populations (such as India and China), which appear on the far right of the histogram.
* The population distribution is highly right-skewed, meaning the majority of countries have low to medium populations, while only a small number have very high populations.
* There is a large variation in population sizes across countries, ranging from countries with a few thousand people to countries with over one billion people.
* The histogram indicates that countries are not evenly distributed by population, with most nations falling into the lower population ranges and only a handful in the highest population range.

```python
fig = px.scatter(
    df,
    x='Land Area (Km²)',
    y='Population (2024)',
    hover_name='Country',
    title='Population vs Area',
    labels={
        'Land Area (Km²)': 'Land Area (Km²)',
        'Population (2024)': 'Population (2024)'
    }
)

fig.update_layout(
    template='plotly_white',
    xaxis_title='Land Area (Km²)',
    yaxis_title='Population (2024)'
)

fig.show()
```

## Insights:
* There is no strong linear relationship between land area and population. Countries with larger land areas do not necessarily have larger populations.
* A few countries are clear outliers, having extremely high populations compared to the rest (such as India and China), making them stand out on the graph.
* Many countries have moderate or small populations regardless of their land area, showing that population depends on multiple factors, not just size.
* Countries with very large land areas can still have relatively low populations (for example, Canada and Australia), indicating low population density.
* The graph shows that land area alone cannot predict population size. Economic opportunities, climate, resources, and urbanization also play a significant role in determining a country's population

```python
fig = px.box(
    df,
    y='Population (2024)',
    title='Box Plot of Population (Outliers)',
    points='outliers'
)

fig.update_layout(
    yaxis_title='Population (2024)',
    template='plotly_white'
)

fig.show()
```

## Insights
* The box plot shows several outliers, meaning a few countries have populations that are much higher than the rest of the world.
* Most countries have relatively low to medium populations, as the main box is concentrated near the lower end of the population scale.
* The highest outlier represents countries with extremely large populations (such as India and China), which are significantly different from other countries.
* The population data is positively (right) skewed, indicating that only a small number of countries have very high populations while the majority have much smaller populations.
* The graph highlights the unequal distribution of population across countries, showing that population is concentrated in only a few nations, whereas most countries have comparatively smaller populations.

```python
import plotly.express as px

fig = px.treemap(
    df,
    path=['Country'],
    values='Population (2024)',
    title='Population by Country'
)

fig.update_layout(
    template='plotly_white'
)

fig.show()
```

## Insights
* India has the largest population, as shown by the biggest rectangle in the treemap, followed closely by China.
* The United States, Indonesia, Pakistan, Brazil, and Nigeria also have large populations, but their blocks are much smaller than India and China.
* Most countries occupy smaller rectangles, indicating that the majority of countries have comparatively lower populations.
* The size of each rectangle is directly proportional to the country's population—larger blocks represent higher populations, while smaller blocks represent lower populations.
* The treemap clearly highlights the unequal distribution of the world's population, showing that a few countries account for a large share of the global population, while many countries contribute a much smaller share.

```python
plt.figure(figsize=(10,6))
plt.scatter(df['Land Area (Km²)'], df['density_per_km_square'], alpha=0.7)
plt.xscale('log')
plt.yscale('log')
plt.title("Density vs Area")
plt.xlabel("Land Area (Km²)")
plt.ylabel("Density (P/Km²)")
plt.grid(True)

plt.show()
```

## Insights
* There is no strong relationship between land area and population density. Countries with large land areas can have either low or high population density.
* Most countries are clustered in the middle of the graph, indicating that many countries have moderate land areas and moderate population densities.
* A few countries have very high population densities, appearing at the top of the graph. These are densely populated countries with many people living in a small area.
* Countries with very large land areas generally tend to have lower population densities, as their populations are spread over a much larger area.
* The logarithmic (log) scale on both axes improves the visualization, making it easier to compare countries with very different land areas and population densities without overcrowding the graph
